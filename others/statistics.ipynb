```python
data = [5, 9, 8, 12, 6, 10, 6, 8]
```

# 2. Average (media)

```python
avg = sum(data) / len(data)
avg
```

# 3. Diferentele la patrat a tuturor elementele

```python
col2 =[f"{i} - {avg} = {i - avg}" for i in data]

col3 = [f"{(i - avg)} ** 2 = {(i - avg) ** 2}" for i in data]

df = pd.DataFrame({"data": data, "Patrat": col2,"Rezultat" :col3})
df
```

# 4. Suma diferentelor la patrat

```python
desc = ""
suma = 0

for i in data:
    suma += (i - avg) ** 2
    desc += ( "+" * bool(desc) +  f"{(i - avg) ** 2}")

desc += f" = {suma}"
desc
```

# 5. Varianța (variance) 

```python
variance = suma / len(data)
variance
```

```python
np.var(data)
```

# 6. Standard variance

```python
np.sqrt(variance)
```

# Dispersie (variabilitate) – pe pași mici

**Ideea:** măsurile de dispersie îți spun *cât de împrăștiate* sunt valorile (nu doar unde e „centrul”).

Începem cu cele mai ușoare:
- **min** și **max**
- **range** = \(\max - \min\)

Apoi trecem la măsuri mai „robuste” (care nu se strică ușor de la 1–2 valori extreme):
- **cuartile** (Q1, mediană, Q3)
- **IQR** = \(Q3 - Q1\)
- **MAD** (Median Absolute Deviation)

La final, o măsură utilă când compari serii cu scări diferite:
- **CV** (coeficient de variație) = \(\sigma / \mu\)


```python
speed_s = pd.Series(speed)

# 1) min și max
_min = speed_s.min()
_max = speed_s.max()

print("min:", _min)
print("max:", _max)
print("range = max - min:", _max - _min)
```

## 2) Cuartile + IQR (pas cu pas)

**Cuartilele** împart datele sortate în 4 părți.
- **Q1 (25%)**: 25% din valori sunt \(\le Q1\)
- **Q2 (50%)**: mediana
- **Q3 (75%)**: 75% din valori sunt \(\le Q3\)

**IQR** (InterQuartile Range) = \(Q3 - Q1\). Este o măsură de dispersie care ignoră cozile extreme.


```python
lower = q1 - 1.5 * iqr
upper = q3 + 1.5 * iqr
outliers = speed_s[(speed_s < lower) | (speed_s > upper)]

print(f"IQR bounds: [{lower:.2f}, {upper:.2f}]")
print("Outlieri:")
print(outliers.to_list() if not outliers.empty else "(niciun outlier după regula IQR)")

plt.figure(figsize=(8, 2.2))
plt.boxplot(speed_s, vert=False, showmeans=True)
plt.title("Boxplot pentru 'speed' (cu mean și outlieri)")
plt.xlabel("speed")
plt.grid(axis='x', alpha=0.2)
plt.show()
```

## 3) Outlieri cu regula IQR (explicat simplu)

Regulă practică:
- **Lower bound** = \(Q1 - 1.5 \cdot IQR\)
- **Upper bound** = \(Q3 + 1.5 \cdot IQR\)

Tot ce este în afara intervalului \([lower, upper]\) se consideră **outlier**.

Important: outlier ≠ „greșit”. Poate fi o valoare reală, doar rară.


```python
lower = q1 - 1.5 * iqr
upper = q3 + 1.5 * iqr
outliers = speed_s[(speed_s < lower) | (speed_s > upper)]

print(f"lower = {lower:.2f}")
print(f"upper = {upper:.2f}")
print("outlieri:", outliers.to_list() if not outliers.empty else "(niciun outlier)")

# Boxplot (rezumă Q1, mediană, Q3 + outlieri)
plt.figure(figsize=(8, 2.2))
plt.boxplot(speed_s, vert=False, showmeans=True)
plt.title("Boxplot pentru 'speed'")
plt.xlabel("speed")
plt.grid(axis='x', alpha=0.2)
plt.show()
```

## 4) Distribuție (cum arată datele): histogramă

Histogramă = „împărțim axa în intervale” și numărăm câte valori cad în fiecare interval.

O regulă simplă de citire:
- dacă histograma e „îngrămădită” într-o zonă → variabilitate mică
- dacă e „împrăștiată” → variabilitate mai mare


```python
# Histogramă pentru 'speed' (cu mean și mediană)
mean = speed_s.mean()
median = speed_s.median()

plt.figure(figsize=(8, 4))
plt.hist(speed_s, bins='auto', edgecolor='black', alpha=0.85)
plt.axvline(mean, color='red', linestyle='--', label=f"mean={mean:.2f}")
plt.axvline(median, color='green', linestyle='--', label=f"median={median:.2f}")
plt.title("Histogramă pentru 'speed'")
plt.xlabel("speed")
plt.ylabel("count")
plt.legend()
plt.grid(alpha=0.2)
plt.show()
```

## 5) ECDF (opțional) – „procentul de valori ≤ x”

ECDF e ca o „numărătoare cumulată”:
- alegi un \(x\)
- te uiți ce proporție din valori sunt \(\le x\)

E utilă când vrei să citești rapid percentile (ex: unde e 80%).


```python
x = np.sort(speed_s.values)
y = np.arange(1, len(x) + 1) / len(x)

plt.figure(figsize=(7, 4))
plt.step(x, y, where='post')
plt.title("ECDF pentru 'speed'")
plt.xlabel("speed")
plt.ylabel("proporție ≤ x")
plt.grid(alpha=0.2)
plt.show()
```

## 6) Z-score (opțional) – „cât de departe e o valoare de medie”

Definiție:
- \(z = (x - \mu) / \sigma\)

Interpretare simplă:
- \(|z|\) mic (ex: sub 1) → valoare „normală”
- \(|z|\) mare (ex: peste 2 sau 3) → valoare „atipică”


```python
std_sample = speed_s.std(ddof=1)
z = (speed_s - mean) / std_sample

pd.DataFrame({"speed": speed_s, "z": z}).sort_values("z")
```

```python
plt.figure(figsize=(10, 3))
plt.stem(speed_s.index, z, basefmt=" ")
plt.axhline(0, color='black', linewidth=1)
plt.axhline(2, color='red', linestyle='--', alpha=0.7, label='z=±2')
plt.axhline(-2, color='red', linestyle='--', alpha=0.7)
plt.title("Z-score pentru fiecare observație din 'speed'")
plt.xlabel("index")
plt.ylabel("z")
plt.legend()
plt.grid(alpha=0.2)
plt.show()
```
