```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from sklearn.datasets import load_iris
iris = load_iris(as_frame=True)
# type(iris)
iris.keys()
```

```python
iris = pd.concat([iris.data, iris.target], axis=1)
iris
```

```python
iris.keys()
```

```python
# iris.target_names
```

```python
iris["species"] = iris["target"].map({k:v for k, v in enumerate(['setosa', 'versicolor', 'virginica'])})
iris
```
